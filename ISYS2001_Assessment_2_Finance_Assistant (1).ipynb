```python
def calculate_total_spending(transactions):
    total = 0

    for transaction in transactions:
        total += transaction["amount"]

    return total
```

```python
def calculate_spending_by_category(transactions):
    spending = {}

    for transaction in transactions:
        category = transaction["category"]
        amount = transaction["amount"]

        if category not in spending:
            spending[category] = 0

        spending[category] += amount

    return spending
```

```python
def compare_with_budget(total_spending, budget):
    remaining = budget - total_spending

    if remaining > 0:
        return f"You are $ {remaining} under your budget."
    elif remaining == 0:
        return "You have exactly reached your budget."
    else:
        return f"You are $ {abs(remaining)} over your budget."
```

```python
# Final Over-Budget Test

test_transactions = [
    {"category": "Food", "amount": 400},
    {"category": "Transport", "amount": 300},
    {"category": "Entertainment", "amount": 500}
]

test_budget = 1000

test_total = calculate_total_spending(test_transactions)
test_categories = calculate_spending_by_category(test_transactions)

print("Total spending:", test_total)
print("Budget:", test_budget)
print(compare_with_budget(test_total, test_budget))
print("Spending by category:", test_categories)

assert test_total == 1200
assert "over your budget" in compare_with_budget(test_total, test_budget)

print("\nOver-budget test passed successfully.")
```

```python
def find_highest_spending_category(spending_by_category):
    highest_category = max(
        spending_by_category,
        key=spending_by_category.get
    )
    highest_amount = spending_by_category[highest_category]

    return highest_category, highest_amount
```

```python
def validate_budget(budget):
    try:
        budget = float(budget)

        if budget < 0:
            return False, "Budget cannot be negative."

        return True, budget

    except (ValueError, TypeError):
        return False, "Please enter a valid number for the budget."


# Test the validation
valid, result = validate_budget(budget)

if valid:
    print("Valid budget:", result)
else:
    print("Error:", result)
```

```python
def create_financial_summary(transactions, budget):
    total_spending = calculate_total_spending(transactions)
    spending_by_category = calculate_spending_by_category(transactions)
    budget_message = compare_with_budget(total_spending, budget)
    highest_category, highest_amount = find_highest_spending_category(
        spending_by_category
    )

    print("===== FINANCIAL SUMMARY =====")
    print("Total spending: $", total_spending)
    print("Budget: $", budget)
    print(budget_message)
    print()
    print("Spending by category:")

    for category, amount in spending_by_category.items():
        print("-", category, ": $", amount)

    print()
    print("Highest spending category:", highest_category)
    print("Highest spending amount: $", highest_amount)


create_financial_summary(transactions, budget)
```

```python
def generate_finance_advice(total_spending, budget, highest_category, highest_amount):
    if total_spending > budget:
        advice = (
            f"You are over your budget by ${total_spending - budget:.2f}. "
            f"Consider reducing spending in {highest_category}, "
            f"which is your highest spending category at ${highest_amount:.2f}."
        )

    elif total_spending == budget:
        advice = (
            f"You have reached your full budget. "
            f"Keep an eye on {highest_category}, "
            f"your highest spending category."
        )

    else:
        remaining = budget - total_spending
        advice = (
            f"You are ${remaining:.2f} under your budget. "
            f"Your highest spending category is {highest_category} "
            f"at ${highest_amount:.2f}."
        )

    return advice


# Calculate the values needed for the advice
total_spending = calculate_total_spending(transactions)

spending_by_category = calculate_spending_by_category(transactions)

highest_category, highest_amount = find_highest_spending_category(
    spending_by_category
)

# Generate financial advice
advice = generate_finance_advice(
    total_spending,
    budget,
    highest_category,
    highest_amount
)

print("Finance Advice:")
print(advice)
```

```python
# Finance assistant persona

SYSTEM_INSTRUCTION = """
You are a helpful and practical Personal Spending & Budget Assistant.

Your role is to help the user understand their personal spending and budget.

When financial transaction data is provided:
- Use the user's provided financial data when giving analysis.
- Explain spending clearly and simply.
- Identify useful patterns in the user's spending.
- Consider the user's budget when discussing their spending.
- Do not invent financial data.
- If the user's data is missing or unclear, ask for the information you need.
- Give practical budgeting guidance based on the available data.

If the user asks something unrelated to personal finance, politely explain that you are designed to help with spending, budgeting and financial analysis.

Do not claim to be a financial adviser or provide professional financial advice.
"""

print("Finance assistant persona created successfully.")
```

```python
def ask_finance_assistant(user_question, transactions, budget):

    total_spending = calculate_total_spending(transactions)
    spending_by_category = calculate_spending_by_category(transactions)

    if spending_by_category:
        highest_category, highest_amount = find_highest_spending_category(
            spending_by_category
        )
    else:
        highest_category = "None"
        highest_amount = 0

    prompt = f"""
Financial data:
Budget: ${budget:.2f}
Total spending: ${total_spending:.2f}
Spending by category: {spending_by_category}
Highest spending category: {highest_category} (${highest_amount:.2f})

User question: {user_question}

Answer using only the financial data provided.
Do not invent financial information.
Keep the answer clear and practical.
"""

    response = client.models.generate_content(
        model="gemini-flash-latest",
        contents=prompt,
        config={
            "system_instruction": SYSTEM_INSTRUCTION
        }
    )

    return response.text
```

```python
# Custom analytical tool: Budget Health Analyzer

def analyze_budget_health(total_spending, budget):
    try:
        # Check that the budget is valid
        budget = float(budget)
        total_spending = float(total_spending)

        if budget <= 0:
            return "Invalid budget: budget must be greater than zero."

        if total_spending < 0:
            return "Invalid spending: spending cannot be negative."

        # Calculate percentage of budget used
        percentage_used = (total_spending / budget) * 100

        # Classify the budget health
        if percentage_used < 75:
            status = "Healthy"
        elif percentage_used <= 100:
            status = "Watch"
        else:
            status = "Over Budget"

        return (
            f"Budget Health: {status}\n"
            f"Budget used: {percentage_used:.1f}%\n"
            f"Total spending: ${total_spending:.2f}\n"
            f"Budget: ${budget:.2f}"
        )

    except (ValueError, TypeError):
        return "Invalid input: please provide valid numbers."
```

```python
# Add the custom Budget Health analysis to the Gemini assistant

import time

def ask_finance_assistant_with_budget_health(user_question, transactions, budget):

    # Calculate the user's financial information
    total_spending = calculate_total_spending(transactions)
    spending_by_category = calculate_spending_by_category(transactions)

    # Find the highest spending category
    if spending_by_category:
        highest_category, highest_amount = find_highest_spending_category(
            spending_by_category
        )
    else:
        highest_category = "None"
        highest_amount = 0

    # Run the custom analytical tool
    budget_health = analyze_budget_health(total_spending, budget)

    # Create financial context for Gemini
    financial_context = f"""
User's financial data:

Budget: ${budget:.2f}

Total spending: ${total_spending:.2f}

Spending by category:
{spending_by_category}

Highest spending category:
{highest_category}: ${highest_amount:.2f}

Budget health analysis:
{budget_health}
"""

    # Create the prompt
    prompt = f"""
{financial_context}

User question:
{user_question}

Answer the user's question using the user's provided financial data
and the budget health analysis.

Do not invent financial information.
Keep the response clear, practical and easy to understand.
"""

    # Ask Gemini with retry handling for temporary 503 errors
    for attempt in range(5):
        try:
            response = client.models.generate_content(
                model="gemini-flash-latest",
                contents=prompt,
                config={
                    "system_instruction": SYSTEM_INSTRUCTION
                }
            )

            return response.text

        except Exception as error:
            if "503" in str(error) and attempt < 4:
                print(
                    f"Gemini temporarily busy. "
                    f"Waiting 10 seconds... Retry {attempt + 1}/4"
                )
                time.sleep(10)
            else:
                raise
```

```python
def parse_transactions_input(spending_text):
    transactions = []

    try:
        lines = spending_text.strip().split("\n")

        for line in lines:
            if not line.strip():
                continue

            category, amount = line.split(":", 1)

            category = category.strip()

            if not category:
                return None, "Please provide a spending category."

            amount = float(
                amount.strip()
                .replace("$", "")
                .replace(",", "")
            )

            if amount < 0:
                return None, "Spending amounts cannot be negative."

            transactions.append({
                "category": category,
                "amount": amount
            })

        if not transactions:
            return None, "Please provide at least one spending category."

        return transactions, None

    except (ValueError, TypeError):
        return None, "Please enter spending data in the format: Category: amount"
```

```python
import time

def finance_chat(spending_text, user_budget, user_question):

    if not user_question or not user_question.strip():
        return "Please enter a finance-related question."

    try:
        budget = float(str(user_budget).strip())
    except (ValueError, TypeError, AttributeError):
        return "Please enter a valid numerical budget."

    if budget <= 0:
        return "Please enter a budget greater than zero."

    transactions, error_message = parse_transactions_input(spending_text)

    if error_message:
        return error_message

    # Calculate financial results using the user's data
    total_spending = calculate_total_spending(transactions)

    # Use the custom Budget Health Analyzer
    budget_health = analyze_budget_health(total_spending, budget)

    # Ask Gemini to answer the user's question using the financial data
    for attempt in range(3):
        try:
            answer = ask_finance_assistant_with_budget_health(
                user_question,
                transactions,
                budget
            )

            return (
                "===== BUDGET HEALTH =====\n"
                + budget_health
                + "\n\n"
                "===== FINANCE ASSISTANT =====\n"
                + answer
            )

        except Exception as error:
            error_message = str(error)

            if "503" in error_message and attempt < 2:
                time.sleep(3)
                continue

            return (
                "===== BUDGET HEALTH =====\n"
                + budget_health
                + "\n\n"
                "Gemini is temporarily unavailable. "
                "Please try your question again later."
            )
```

```python
import gradio as gr
```

```python
finance_app = gr.Interface(
    fn=finance_chat,
    inputs=[
        gr.Textbox(
            label="Your Spending Data",
            placeholder="Food: 500\nTransport: 300\nEntertainment: 300\nGroceries: 1000",
            lines=6
        ),
        gr.Number(
            label="Monthly Budget",
            value=3500
        ),
        gr.Textbox(
            label="Ask your Finance Assistant",
            placeholder="Example: Where am I spending the most money?",
            lines=3
        )
    ],
    outputs=gr.Textbox(
        label="Finance Assistant Response",
        lines=12
    ),
    title="Personal Spending & Budget Assistant",
    description=(
        "Enter your spending data and monthly budget, then ask questions "
        "about your spending. The assistant analyses the financial data "
        "you provide."
    )
)
```

## Testing and Debugging: Finding and Fixing a Problem

During testing, I checked how the finance assistant behaved when no transaction data was provided.

The original version of the program produced an error because it tried to find the highest spending category when the transaction list was empty.

This showed that the program needed better handling for empty transaction data.

I fixed the problem by adding a check for an empty transaction list before performing the calculations.

After making the change, I tested the program again with an empty transaction list. The program no longer crashed and instead displayed:

**"No transaction data was provided."**

This confirmed that the problem had been fixed and that the program could handle missing transaction data more safely.

```python
# Fix: Handle empty transaction data safely

def create_financial_summary(transactions, budget):
    if not transactions:
        print("No transaction data was provided.")
        return

    total_spending = calculate_total_spending(transactions)
    spending_by_category = calculate_spending_by_category(transactions)
    budget_message = compare_with_budget(total_spending, budget)
    highest_category, highest_amount = find_highest_spending_category(
        spending_by_category
    )

    print("===== FINANCIAL SUMMARY =====")
    print("Total spending: $", total_spending)
    print("Budget: $", budget)
    print(budget_message)
    print()
    print("Spending by category:")

    for category, amount in spending_by_category.items():
        print("-", category, ": $", amount)

    print()
    print("Highest spending category:", highest_category)
    print("Highest spending amount: $", highest_amount)
```
